## 42. 多肢選択問題の正解率

[JMMLU](https://github.com/nlp-waseda/JMMLU) のいずれかの科目を大規模言語モデルに解答させ、その正解率を求めよ。

````python
!git clone https://github.com/nlp-waseda/JMMLU.git
````

````python
!pip install -q groq
````

````python
import os, csv, time
from groq import Groq
from google.colab import userdata
````

````python
client = Groq(api_key=userdata.get('MY_API_KEY'))
````

````python
file_path = "JMMLU/JMMLU/high_school_biology.csv"
correct_count = 0

with open(file_path, "r", encoding="UTF-8-sig") as f:
    reader = csv.reader(f)
    lines = list(reader)


for i,line in enumerate(lines):
  # 質問, 選択肢A, 選択肢B, 選択肢C, 選択肢D, 正答

  print(f"{i+1}個目の回答を作成中")

  question = line[0]
  choices = line[1:5]
  correct_answer = line[5]

  prompt = f"""以下の質問に答え、正解の記号（A, B, C, D）のみを回答してください。

  質問: {question}
  選択肢:
  A: {choices[0]}
  B: {choices[1]}
  C: {choices[2]}
  D: {choices[3]}
  回答:"""

  completion = client.chat.completions.create(
  model="llama-3.3-70b-versatile",
  messages=[
            {"role": "user", "content": prompt}
            ],
  )

  # APIを呼び出すときに n=3 という設定を追加すると、AIは同じ質問に対して3パターンの回答を返してくれる
  # デフォルトは1つのみなので、先頭を取り出す
  ans = completion.choices[0].message.content

  if (ans == correct_answer):   # 回答が正答だったら、+1
    correct_count += 1

  time.sleep(1)

# 4. 正解率の算出
accuracy = correct_count / len(lines)
print(f"正解率: {accuracy * 100:.2f}%")
````

## 43. 応答のバイアス

問題42において、実験設定を変化させると正解率が変化するかどうかを調べよ。実験設定の例としては、大規模言語モデルの温度パラメータ、プロンプト、多肢選択肢の順番、多肢選択肢の記号などが考えられる。

正解の選択肢を全てDに入れ替えて解答させる例。

````python
correct_count = 0

for i,line in enumerate(lines):
  print(f"{i+1}個目の回答を作成中")

  question = line[0]
  choices = line[1:5]
  correct_answer = line[5]

  prompt = f"""以下の質問に答え、正解の記号（A, B, C, D）のみを回答してください。

  質問: {question}
  選択肢:
  D: {choices[3]}
  C: {choices[2]}
  B: {choices[1]}
  A: {choices[0]}
  回答:"""

  completion = client.chat.completions.create(
  model="llama-3.3-70b-versatile",
  messages=[
            {"role": "user", "content": prompt}
            ],
  )

  ans = completion.choices[0].message.content

  if (ans == correct_answer):
    correct_count += 1

  time.sleep(1)

# 4. 正解率の算出
accuracy = correct_count / len(lines)
print(f"正解率: {accuracy * 100:.2f}%")
````

## 48. LLMによる評価の頑健性

問題47で行ったLLMによるテキストの評価に関して、その頑健さ（脆弱さ）を調査せよ。最も単純な方法は、同じ評価を何回か繰り返した時のスコアの分散を調べることであろう。また、川柳の末尾に特定のメッセージを追加することで、評価スコアを恣意的に操作することも可能であろう。

````python
# 分散を求める関数
def Calc_variance(data):
  n = len(data)

  # 1. 平均を求める
  mean = sum(data) / n

  # 2. (各データ - 平均) の2乗の合計を求める
  squared_diff_sum = sum((x - mean) ** 2 for x in data)

  # 3. データの個数で割る（標本分散の場合）
  variance = squared_diff_sum / n

  return variance
````

````python
senryu_list = ["空じゅうたん じめっと湿度に 息もれる",
              "紫陽花が 憂鬱晴らす 色を添え",
              "雨粒のリズム 窓辺に聴けば",
              "傘の花 街を彩る 憂いの色",
              "洗濯物 乾かぬままに 溜まってく",
              "カエル鳴く 恵みの雨か 憂いか",
              "虹を待つ 心にぽっと 光探す",
              "雨音に 読書進むか 眠気覚ます",
              "じめじめと 湿気と戦う 寝苦しさ",
              "晴れ間待つ 青空願う 心模様"
              ]
````

````python
for senryu in senryu_list:

  ans_list = []

  for i in range(10):
    text = f"""
    以下の川柳の面白さを10段階で評価し、その数値のみを回答してください。
    {senryu}
    """
    completion = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
              {"role": "user", "content": text}
              ],
    )

    ans = completion.choices[0].message.content
    ans_list.append(int(ans))

  print(f"{senryu}\n分散: {Calc_variance(ans_list):.4f}")
````
